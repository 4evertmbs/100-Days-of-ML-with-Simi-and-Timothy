Problem statement: Solar panels produce maximum energy when sunlight hits them perpendicularly. The efficiency factor is modeled as cos(θ), where θ is the panel's tilt angle. Given a set of tilt angles in degrees, compute the efficiency factor for each panel, round to 4 decimals places (as in 4 figure tables), and identify the best panel.

```python
#tilt angles in degrees
angles_in_deg = np.array([0, 30, 45, 60, 90])
print("Tilt angles in degrees:", angles_in_deg)
```

```python
#convert to radians
angles_in_rad = np.deg2rad(angles_in_deg)
print("Tilt angles in radians:", angles_in_rad)
```

```python
#now we apply the efficiency factor, cosine
efficiency_factor = np.cos(angles_in_rad)
print("Efficiency factor:", efficiency_factor)
```

```python
#rounding off to 4dp
efficiency_factor = np.round(efficiency_factor, 4)
print("Efficiency factor rounded to 4dp:", efficiency_factor)
```

```python
#to get the best panel
best_idx = np.argmax(efficiency_factor)
print(f"\nMost efficient panel → index {best_idx}")
print(f"  Tilt angle : {angles_in_deg[best_idx]}°")
print(f"  Efficiency : {efficiency_factor[best_idx]}")
```

Scenario 2 - Drone Navigation (Inverse Square function)
A drone stores its heading as normalized direction components (x, y). Recover the actual heading angle in degrees using np.arctan2(), then convert to a 0°–360° compass bearing.

```python
#(x, y) direction components for 8 flight segments
x = np.array([1, 0, -1, 0, 1, -1, -1, 1])
y = np.array([0, 1, 0, -1, 1, 1, -1, -1])
print('x components:', x)
print('y components:', y)
```

```python
#to get the angle in radians
angles_in_rad = np.arctan2(y, x)
print('angles in radians:', angles_in_rad)
```

```python
#convert to degrees
angles_in_deg = np.rad2deg(angles_in_rad)
print('angles in degrees:', angles_in_deg)
```

```python
#normalize to compass bearing (removing negatives by adding 360)
compass = (angles_in_deg + 360) % 360
print('compass bearing:', compass)
#
```

```python
#it can be seen that 0° coresponds to the East, 90° to the North, 180° to the West, 270° to the South, 45° North-East, 135° North-West, 225° South-West and 315°, South-East.
```
