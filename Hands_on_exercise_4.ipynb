Exercise 1

```python
import numpy as np
```

```python
#create customer data
customer_data = np.array([
    [250000, 120000, 18, 0.40],
    [180000,  90000, 12, 0.55],
    [320000, 150000, 25, 0.30],
    [210000, 100000, 15, 0.65],
    [400000, 200000, 30, 0.25]
])
print("customer_data:", customer_data)
print("\ncustomer_data.shape:", customer_data.shape)
```

```python
scale = np.array([.001, .001, .1, 100])
print("scale:", scale)
print("scale.shape:", scale.shape)
```

```python
#we multiply each column by its scale factor
scaled_data = customer_data * scale
print("scaled_data:\n", scaled_data)
print("\nscaled_data.shape:", scaled_data.shape)
```

```python
#to show that Broadcasting is happening
# a (5, 4) array can be multiplied by a (4, ) array because as it is in Numpy, the columns and rows align from right to left.
#scale's dimension is 4 and customer_date rightmost dimension is 4 though it has 5 rows. the missing dimension of scale is assumed to be one so it becomes (1, 4)
# hence, (5, 4) and (1, 4) are compatible.
```

```python
#results interpretation
```

```python
#.001 affects Monthly income and Monthly spending converting them to thousands
#.1 affects Number of transactions shrinking them from 10 to 1
#100 affects Credit Utilization and it converts 1 unit to 100 units
#The output stays (5, 4) because broadcasting adjusts the smaller array to fit the larger one without changing its shape. Each customer still has 4 features.
```

```python
#manual verification of a customer
```

```python
first_customer = customer_data[0]
print("first_customer:", first_customer)
```

```python
manual_scaled = first_customer * scale
print("Manual scaled:", manual_scaled)
```

```python
print("Broadcast scaled:", scaled_data[0])
print("Match:", np.allclose(manual_scaled, scaled_data[0]))
```

Exercise 2

```python
waiting_times =  np.array([
    [45, 30, 20],
    [60, 40, 25],
    [35, 50, 30],
    [80, 45, 35],
    [55, 35, 20],
    [40, 60, 40]
])
target_wait = np.array([30, 35, 25])
print("waiting_times:\n", waiting_times)
print("waiting_times.shape:", waiting_times.shape)
print("\ntarget_wait:", target_wait)
print("target_wait.shape:", target_wait.shape)
```

```python
difference = waiting_times - target_wait
print("difference:\n", difference)
print("shape:", difference.shape)
```

```python
#15 means a patient waited 15 mins longer than normal
#-5 means a patient waited 5 mins less than normal
#0 means a patient doesn't have to wait, i.e exactly on target
```

```python
#Identifying patients above target
above_target = difference > 0
print("above_target:\n", above_target)
print("shape:", above_target.shape)
```

```python
# True = the patient waited longer than expected
# False = the patient waited less than or equal to the target
```

```python
#percentage difference
percentage_difference = (difference / target_wait) * 100
print("percentage_difference:\n", percentage_difference)
print("shape:", percentage_difference.shape)
```

```python
#largest delay
max_delay = np.max(difference)
print("largest delay:", max_delay, "mins")
```

```python
#position of max delay
row_idx, col_idx = np.where(difference == max_delay)
print("row_idx:", row_idx)
print("col_idx:", col_idx)
```

```python
#patient 4 with row index 3 waited 50mins above the 30mins targer
```

```python
#Bonus Target
#average difference per department
avg_diff_per_dept = np.mean(difference, axis=0)
print("avg_diff_per_dept:", avg_diff_per_dept)
departments = ["Emergency", "Outpatient", "Laboratory"]
for dept, avg_diff in zip(departments, avg_diff_per_dept):
    print(f"Average difference for {dept}: {avg_diff:.2f} mins")
```

```python
#Emergency performs furthest away from its target
#Outpatient performs better with a time of 8.33mins above its target
#Laboratory is closest with a time of 3.33mins
```

```python
#understanding the image shape
np.random.seed(42)
images = np.random.randint(0, 256, size=(8, 3, 4, 4))
print("images.shape:", images.shape)
```

```python
#Axis 0 has size 8, number of images in the batch
#Axis 1 has size 3, colour channels (R, G, B)
#Axis 2 and 3 haves size 4, image height and width respectively
```

```python
#inspect the channel scale
channel_scale = np.array([0.8, 1.0, 1.2])
print('channel_scale:', channel_scale)
print('shape:', channel_scale.shape)
```

```python
#reshape the channel
channel_scale_reshaped = channel_scale.reshape(1, 3, 1, 1)
print('channel_scale_reshaped:', channel_scale_reshaped)
print('shape:', channel_scale_reshaped.shape)
#
```

```python
#applying broadcasting
adjusted_images = images * channel_scale_reshaped
print("adjusted_images.shape:", adjusted_images.shape)
```

```python
#why it works
#Axis 0 is compatible because 1 broadcasts to 8
#Axis 1 is an exact match as 3 broadcasts to 3
#Axis 2 and 3 are compatible because 1 broadcasts to 4
```

```python
#verifying 1 pixel
pixel_rgb = images[0, :, 0, 0]
print("Original RGB pixel:", pixel_rgb)
#manual scaling
manual_scaled_pixel = pixel_rgb * channel_scale
print("Manual scaled RGB pixel:", manual_scaled_pixel)
```

```python
#broadcast result
broadcast_scaled_pixel = adjusted_images[0, :, 0, 0]
print("Broadcast scaled RGB pixel:", broadcast_scaled_pixel)
print("Match:", np.allclose(manual_scaled_pixel, broadcast_scaled_pixel))
```

```python
#Bonus challenge
channel_mean = np.array([120, 130, 140]).reshape(1, 3, 1, 1)
channel_std = np.array([50, 40, 30]).reshape(1, 3, 1, 1)

normalized = (images - channel_mean) / channel_std
print("normalized.shape:", normalized.shape)
```

```python
#channel_mean and channel_std have one value for each channel. Reshaping them to (1, 3, 1, 1) lets NumPy apply them across all 8 images and their pixels at once. This makes the code simpler, faster, and avoids loops.
```
